```python
from faker import Faker
import pandas as pd
import numpy as np
import json
import random
from datetime import datetime

fake = Faker()

# Lots fixes de villes 
Transporteurs= ["ColisExpress", "RapidTrans", "DDPP", "LivrHome", "KKExpress", "RelaisMaison", "LaBP"]

#COMMANDES = [f"CAPT-{i:03d}" for i in range(1, 16)]



datajson = [
  {
    "commande_id": "CMD-10234",
    "client_id": "CLI-5521",
    "ville_livraison": "Lyon",
    "date_commande": "2024-09-12T10:03:00",
    "date_livraison_prevue": "2024-09-13T18:00:00",
    "date_livraison_reelle": "2024-09-14T09:20:00",
    "poids_kg": 2.3,
    "transporteur": "ColisExpress",
    "distance_km": 14.5,
    "nb_articles": 3,
    "statut": "livre"
  },
  {
    "commande_id": "CMD-10235",
    "client_id": "CLI-8890",
    "ville_livraison": None,
    "date_commande": "2024-09-12T11:45:00",
    "date_livraison_prevue": "2024-09-13T18:00:00",
    "date_livraison_reelle": None,
    "poids_kg": None,
    "transporteur": "RapidTrans",
    "distance_km": 8.2,
    "nb_articles": 1,
    "statut": "en_cours"
  }
]

def valeur_ou_manquante(valeur, proba_manquant=0.08):
    return None if random.random() < proba_manquant else valeur



for i in range(1000):
    start = fake.date_time_between(start_date="-1000d", end_date="now").isoformat()
    livprev = fake.date_time_between(start_date=pd.to_datetime(start), end_date="now").isoformat()
    livrel= fake.date_time_between(start_date=pd.to_datetime(livprev), end_date="now").isoformat()

    nb_article = fake.random_int(min=1, max=120)
    poids = round(fake.pyfloat(min_value=-5, max_value=135), 1)
    distance = round(fake.pyfloat(min_value=0, max_value=80), 1)

    if random.random() < 0.1:
        poids = round(random.choice([-99.9, 150.0, 999.0]), 1)
    if random.random() < 0.1:
        nb_article = random.choice([-1, 50000, 99999])
    if random.random() < 0.1:
        distance = random.choice([-5.0, 9999.9])

    obj = {
        "commande_id": valeur_ou_manquante("CMD-" + str(fake.random_int(min=20, max=10000))),
        "client_id": valeur_ou_manquante("CLI-" + str(fake.random_int(min=20, max=10000))),
        "ville_livraison": valeur_ou_manquante(fake.city()),
        "date_commande": start,
        "date_livraison_prevue": valeur_ou_manquante(livprev),
        "date_livraison_reelle": valeur_ou_manquante(livrel),
        "poids_kg": poids,
        "transporteur": valeur_ou_manquante(fake.random_element(elements=Transporteurs)),
        "distance_km": distance,
        "nb_articles": nb_article,
        "statut": fake.random_element(elements=("livre", "inconnu", "en_cours"))
    }
    datajson.append(obj)

# --- doublons ---
nb_duplicats = random.randint(12,25)
duplicats = [dict(random.choice(datajson)) for _ in range(nb_duplicats)]
datajson.extend(duplicats)
random.shuffle(datajson)

print(f"Total lignes : {len(datajson)} (dont {nb_duplicats} doublons)")


json_str = json.dumps(datajson, indent=4)
with open("commandes_raw.json", "w") as f:
    f.write(json_str)
```
